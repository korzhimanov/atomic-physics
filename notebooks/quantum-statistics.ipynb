#  Распределение Ферми — Дирака

Среднее число заполнения при фиксированном химическом потенциале $\mu$ и различных температурах:

$$
\bar n(\varepsilon) = \frac{1}{\exp\left(\frac{\varepsilon - \mu}{kT}\right) + 1}
$$

Энергия отложена в единицах $\mu$, температура задаётся отношением $kT/\mu$.

```python
from numpy import exp, linspace
```

```python
x = linspace(0, 2, 2001)  # ε/μ
temperatures = [0.05, 0.15]  # kT/μ
n = {t: 1/(exp((x - 1)/t) + 1) for t in temperatures}
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
styles = ['default', 'seaborn-v0_8-paper', 'seaborn-v0_8-whitegrid', 'seaborn-v0_8-deep']
plt.style.use([s if s in plt.style.available else s.replace('-v0_8', '') for s in styles])
plt.rcParams.update({'axes.labelsize': 12, 'xtick.labelsize': 11, 'ytick.labelsize': 11, 'legend.fontsize': 11})
```

```python
fig = plt.figure(figsize=(4.5, 3))

ax = fig.add_subplot(111)

ax.plot([0, 1, 1, 2], [1, 1, 0, 0], 'k--', label='$T = 0$')
for t in temperatures:
    ax.plot(x, n[t], label=f'$kT = {t:g}\\,\\mu$'.replace('.', '{,}'))

ax.set_xlim(0, 2)
ax.set_ylim(0, 1.1)
ax.set_xticks([0, 1, 2], ['0', '$\\mu$', '$2\\mu$'])
ax.set_yticks([0, 0.5, 1], ['0', '1/2', '1'])

ax.legend()

plt.xlabel('$\\varepsilon$')
plt.ylabel('$\\bar n$')

plt.savefig('../img/fermi-dirac-distribution.svg', bbox_inches='tight')
#plt.show()
```
